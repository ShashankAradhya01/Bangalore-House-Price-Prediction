```python
import re

def convert_sqft(value):
    value = str(value).strip().lower()

    # Convert common area units into square feet
    unit_factors = {
        "sq. meter": 10.7639,
        "sq meter": 10.7639,
        "sq. yards": 9,
        "sq yards": 9,
        "acres": 43560,
        "acre": 43560,
        "cents": 435.6,
        "guntha": 1089,
        "grounds": 2400,
        "perch": 272.25
    }

    for unit, factor in unit_factors.items():
        if unit in value:
            match = re.search(r"\d+(?:\.\d+)?", value)
            return float(match.group()) * factor if match else np.nan

    # Handle ranges such as "2100 - 2850"
    numbers = re.findall(r"\d+(?:\.\d+)?", value)

    if "-" in value and len(numbers) >= 2:
        return (float(numbers[0]) + float(numbers[1])) / 2

    if numbers:
        return float(numbers[0])

    return np.nan


data["total_sqft"] = data["total_sqft"].apply(convert_sqft)

print(data["total_sqft"].describe())
print("Missing square-foot values:", data["total_sqft"].isna().sum())
```

```python
def outlier_report(series):
    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)

    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    flagged = (series < lower) | (series > upper)

    return {
        "Q1": q1,
        "Q3": q3,
        "IQR": iqr,
        "Lower Bound": lower,
        "Upper Bound": upper,
        "Potential Outliers": int(flagged.sum()),
        "Percentage Flagged": round(flagged.mean() * 100, 2)
    }


for column in ["price", "total_sqft", "bhk"]:
    print(f"\n{column.upper()}")
    print(pd.Series(outlier_report(eda[column])).round(2))
```

```python
# Use the EDA dataset created in Stage 3
model_data = eda.copy()

print("Dataset shape:", model_data.shape)
print("Columns:", model_data.columns.tolist())

display(model_data.head())
```

```python
before = len(model_data)

model_data = model_data[
    (model_data["price"] > 0) &
    (model_data["total_sqft"] > 0) &
    (model_data["bhk"] > 0) &
    (model_data["location"].notna()) &
    (model_data["location"].str.strip() != "")
].copy()

print("Records before filtering:", before)
print("Records after filtering:", len(model_data))
print("Records removed:", before - len(model_data))
```

```python
def find_iqr_outliers(df, column):
    q1 = df[column].quantile(0.25)
    q3 = df[column].quantile(0.75)
    iqr = q3 - q1

    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    mask = (df[column] < lower) | (df[column] > upper)

    print(f"\nColumn: {column}")
    print(f"Lower bound: {lower:.2f}")
    print(f"Upper bound: {upper:.2f}")
    print(f"Potential outliers: {mask.sum()}")

    return df.loc[mask]

price_outliers = find_iqr_outliers(model_data, "price")
area_outliers = find_iqr_outliers(model_data, "total_sqft")

display(price_outliers.head(10))
display(area_outliers.head(10))
```

```python
model_data["location"] = (
    model_data["location"]
    .astype(str)
    .str.strip()
)

location_counts = model_data["location"].value_counts()

print("Unique locations:", model_data["location"].nunique())
print("\nMost common locations:")
display(location_counts.head(20))

print("\nLocations with fewer than 10 records:",
      (location_counts < 10).sum())
```

```python
import os
import numpy as np
import pandas as pd
import joblib

from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer

from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.dummy import DummyRegressor

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

print("Machine learning libraries imported successfully!")
```

```python
features = [
    "location",
    "total_sqft",
    "bhk",
    "bath",
    "balcony",
    "area_type"
]

target = "price"

# Keep only the columns needed for this version
training_data = model_data[features + [target]].copy()

# Remove records with invalid target values
training_data = training_data.dropna(subset=[target])
training_data = training_data[training_data[target] > 0]

X = training_data[features]
y = training_data[target]

print("Feature shape:", X.shape)
print("Target shape:", y.shape)

print("\nFeatures:")
print(X.columns.tolist())

display(training_data.head())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training records:", len(X_train))
print("Testing records:", len(X_test))
```

```python
numeric_features = [
    "total_sqft",
    "bhk",
    "bath",
    "balcony"
]

categorical_features = [
    "location",
    "area_type"
]

numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median"))
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(
        handle_unknown="ignore",
        min_frequency=10
    ))
])

preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, numeric_features),
    ("categorical", categorical_pipeline, categorical_features)
])

print("Preprocessing pipeline created!")
```

```python
models = {
    "Baseline": DummyRegressor(strategy="median"),

    "Linear Regression": LinearRegression(),

    "Decision Tree": DecisionTreeRegressor(
        max_depth=15,
        min_samples_leaf=5,
        random_state=42
    ),

    "Random Forest": RandomForestRegressor(
        n_estimators=200,
        min_samples_leaf=2,
        random_state=42,
        n_jobs=-1
    )
}

results = []

for name, regressor in models.items():

    pipeline = Pipeline([
        ("preprocessor", preprocessor),
        ("model", regressor)
    ])

    # Cross-validation uses training data only
    scores = cross_val_score(
        pipeline,
        X_train,
        y_train,
        cv=3,
        scoring="neg_mean_absolute_error",
        n_jobs=1
    )

    mean_mae = -scores.mean()
    std_mae = scores.std()

    results.append({
        "Model": name,
        "CV MAE (Lakhs)": mean_mae,
        "CV MAE Std": std_mae
    })

    print(
        f"{name}: "
        f"CV MAE = {mean_mae:.2f} lakhs "
        f"(std = {std_mae:.2f})"
    )

results_df = pd.DataFrame(results).sort_values(
    "CV MAE (Lakhs)"
).reset_index(drop=True)

display(results_df.round(3))
```

```python

best_model_name = results_df.loc[0, "Model"]

best_regressor = models[best_model_name]

final_model = Pipeline([
    ("preprocessor", preprocessor),
    ("model", best_regressor)
])

final_model.fit(X_train, y_train)

# Predictions on data the model has not seen
y_pred = final_model.predict(X_test)

mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("Selected model:", best_model_name)
print(f"Test MAE:  {mae:.2f} lakhs")
print(f"Test RMSE: {rmse:.2f} lakhs")
print(f"Test R²:   {r2:.4f}")
```

```python
plt.figure(figsize=(8, 6))

plt.scatter(
    y_test,
    y_pred,
    alpha=0.5
)

lower = min(y_test.min(), y_pred.min())
upper = max(y_test.max(), y_pred.max())

plt.plot(
    [lower, upper],
    [lower, upper],
    linestyle="--"
)

plt.xlabel("Actual Price (Lakhs)")
plt.ylabel("Predicted Price (Lakhs)")
plt.title(f"Actual vs Predicted Prices — {best_model_name}")
plt.tight_layout()
plt.show()
```

```python
comparison = X_test.copy()

comparison["Actual Price"] = y_test
comparison["Predicted Price"] = y_pred

comparison["Absolute Error"] = abs(
    comparison["Actual Price"] -
    comparison["Predicted Price"]
)

display(
    comparison[
        [
            "location",
            "total_sqft",
            "bhk",
            "Actual Price",
            "Predicted Price",
            "Absolute Error"
        ]
    ].head(15).round(2)
)
```
